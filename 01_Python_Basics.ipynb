## Section 2 – Control Flow

### if / elif / else
`if`, `elif`, and `else` help Python make decisions.

* `if` checks the first condition.
* `elif` checks another condition if the first one is false.
* `else` runs when none of the conditions are true.

Only one matching block is executed.

Example

Imagine traffic lights.

* If the light is Green → Go
* Else if the light is Yellow → Slow
* Else → Stop

Python checks each condition one by one. As soon as one condition is true, it performs that action.

````python
# Grading a student's marks
marks = 78

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

### for loop
#### Definition

A for loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.

### Real-Life Example

Imagine a teacher checking attendance for 30 students.

Instead of saying each student's name manually in separate actions, the teacher goes through the class list one student at a time until everyone is checked.

A for loop works the same way—it automatically repeats the same task for every item in a sequence.
### Example

````python
# Print squares of numbers from 1 to 5
for i in range(1, 6):
    print(i, "*", i, "=", i * i)
````

### while loop

#### Definition

A while loop is a control statement that repeats a block of code as long as a specified condition is True.

### Real-Life Example

Imagine filling a water tank.

While the tank is not full, keep filling it.

When the tank becomes full, stop.

The loop continues until the condition becomes False.

### Example

````python
# Countdown using while loop
count = 5

while count > 0:
    print(count)
    count -= 1

print("Liftoff!")
````

## Break
### Definition

The `break` statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

### Real-Life Example

Imagine you're searching for your friend in a crowd.

As soon as you find them, you stop searching.

You don't continue looking through the remaining people.

That's exactly what `break` does.

### Example

````python
# Stop the loop once the target roll number is found
roll_numbers = [101, 102, 103, 104, 105]
target = 103

for roll in roll_numbers:
    if roll == target:
        print("Student found:", roll)
        break
else:
    print("Not in the list")
````

# Continue
## Definition

The `continue` statement is used to skip the current iteration of a loop and continue with the next iteration.

## Real-Life Example

Imagine a teacher checking homework.

One student forgot the homework.

The teacher skips that student and continues checking the rest of the class.

The checking doesn't stop—it simply skips one student.

### Example

````python
# Skip absent students (marked as None) while printing attendance
students = ["Aishu", None, "Divya", None, "Kavi"]

for student in students:
    if student is None:
        continue
    print(student, "is present")
````

# List
### Definition

A list is an ordered, mutable collection that can store multiple items. It allows duplicate values and items can be added, removed, or modified.

Ordered → Items keep their position.
Mutable → Items can be changed after creation.

### Real-Life Example

Think of your shopping list.

Milk
Bread
Eggs
Rice

### You can:

Add a new item.
Remove an item.
Change an item.

Just like a Python list.

````python
# Create a list of students
students = ["Aishu", "Brindha", "Celine", "Divya"]
print(students)

# Add a new student
students.append("Arthi")
print(students)

# Remove a student
students.remove("Celine")
print(students)

# Change a student's name using index
students[1] = "Barathi"
print(students)

# Access a particular student
print(students[0])

# Total number of students
print(len(students))

# Sort in alphabetical order
students.sort()
print(students)
````

## Tuple
### Definition

A tuple is an ordered, immutable collection that stores multiple items. Once created, its values cannot be changed.

Ordered → Items keep their position.
Immutable → Items cannot be modified.

### Real-Life Example

Think of your Date of Birth.

### Example:

15-08-2000

Your date of birth doesn't change.

Similarly, once a tuple is created, its values remain the same.

### Easy way to remember:

List ([]) → Can add, remove, and change items.

Tuple (()) → Can only store, access, count, and search items. It cannot be modified after creation.

### Example:

birth_date = (15, 8, 2000)

````python
# Create tuple and access elements
fruits = ("apple", "banana", "mango", "orange", "papaya")

print(fruits)
print(fruits[0])          # indexing
print(fruits[-1])         # negative indexing
print(fruits[1:4])        # slicing
print(len(fruits))        # length
print("apple" in fruits)  # membership

# concatenation
colours = ("red", "orange", "yellow")
print(fruits + colours)

# tuple unpacking
student = ("Aishu", 24, "Chennai")
name, age, city = student
print(name, age, city)

# count elements
print(fruits.count("orange"))
````

````python
# Immutable - this raises an error
fruits = ("apple", "banana", "mango")
fruits[1] = "orange"
````

## Set
### Definition

A set is an unordered collection of unique items. It automatically removes duplicate values.

Unordered → Items do not have a fixed position.
Unique → Duplicate values are not allowed.

### Real-Life Example

Imagine a classroom.

````python
# only unique items
numbers = {1, 2, 2, 3, 4, 4}
print(numbers)

# adding items
fruits = {"apple", "banana"}
fruits.add("mango")
print(fruits)

# removing items
fruits.remove("banana")
print(fruits)

# union of 2 sets
frt = {"apple", "orange", "banana"}
clr = {"red", "orange", "green"}
print(frt.union(clr))

# intersection of 2 sets
print(frt.intersection(clr))
````

## List Comprehension
### Definition

List comprehension is a short and simple way to create a new list by writing the loop and optional condition in a single line.

### Real-Life Example

Imagine a teacher writing only the names of students who passed an exam.

Instead of writing a long process, the teacher directly creates a new list containing only the passed students.

Similarly, list comprehension creates a new list in one line.

### Example:

````python
marks = [45, 88, 32, 91, 67]

# list of marks that are passing (>=50)
passed = [m for m in marks if m >= 50]
print(passed)

# lengths of words
words = ["cat", "elephant", "dog"]
lengths = [len(word) for word in words]
print(lengths)
````

## Variable-Length Arguments (`*args`, `**kwargs`)
### Definition

Variable-length arguments allow a function to accept any number of arguments.

## Recursive Functions
### Definition

A recursive function is a function that calls itself to solve a problem. It continues calling itself until a stopping condition (called the base case) is reached.

### Real-Life Example

Imagine standing between two mirrors.

You see your reflection repeating again and again.

It continues until the reflections become too small to notice.

Similarly, a recursive function keeps calling itself until it reaches the stopping condition.

### Example

````python
# factorial of a number
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
````

## Generators
### Definition

A generator is a special type of function that produces values one at a time instead of returning all values at once.

### Real-Life Example

Imagine a ticket machine.

It gives one ticket each time you press the button instead of printing all tickets at once.

A generator works the same way.

### Example

````python
def square():
    for i in range(2, 6):
        yield i * i

gen = square()

for num in gen:
    print(num)
````

## yield
### Definition

The `yield` keyword is used inside a generator function to return one value at a time while saving the function's state. The function resumes from where it stopped when called again.

### Real-Life Example

Think of a TV series.

You watch one episode today and continue from the next episode tomorrow.

It doesn't start from Episode 1 every time.

Similarly, `yield` pauses and resumes the function.

### Example

````python
def even():
    for i in range(1, 11):
        if i % 2 == 0:
            yield i

for num in even():
    print(num)
````

## Decorators
### Definition

A decorator is a function that adds extra functionality to another function without changing its original code.

### Real-Life Example

Think of gift wrapping.

The gift remains the same, but wrapping adds decoration.

Similarly, a decorator adds extra behavior to a function.

### Example

````python
def decorator(func):
    def wrapper(a, b):
        print("Adding numbers...")
        func(a, b)
    return wrapper

@decorator
def add(a, b):
    print(a + b)

add(10, 20)
````

## Closures
### Definition

A closure is a function that remembers the variables from its outer function even after the outer function has finished executing.

## Real-Life Example

Imagine your mother gives you a lunch box before you leave for school.

Even after you leave home, you still have the lunch box.

Similarly, the inner function remembers data from the outer function.

### Example

````python
def multiply(x):
    def inner(y):
        print(x * y)

    return inner

result = multiply(5)
result(4)
````

## global and nonlocal
### Definition

`global` allows a function to modify a variable defined outside all functions (global scope).
`nonlocal` allows an inner function to modify a variable in its enclosing (outer) function.

### Real-Life Example

Imagine a company.

`global` → Company-wide announcement.
`nonlocal` → Department-level announcement.

A company announcement affects everyone.

A department announcement affects only that department.

### Example

````python
balance = 1000

def deposit():
    global balance
    balance += 500

deposit()
print(balance)

def outer():
    count = 0
    def inner():
        nonlocal count
        count += 1
        return count
    print(inner())
    print(inner())

outer()
````

## map()
Definition

The `map()` function applies a given function to every item in an iterable (such as a list or tuple) and returns a map object.

Real-Life Example

Imagine a teacher adds 5 bonus marks to every student's score.

Instead of updating each score one by one, the teacher applies the same rule to all students.

### Example

````python
marks = [40, 55, 62, 70]

# add 5 bonus marks to every score
bonus_marks = list(map(lambda m: m + 5, marks))
print(bonus_marks)

fruits = ["apple", "banana", "mango"]
result = list(map(str.upper, fruits))
print(result)
````

## all()
### Definition

The `all()` function returns True only if every element in an iterable is True. If even one element is False, it returns False.

### Real-Life Example

Imagine checking whether all students submitted their assignments.

### Example

````python
numbers = [1, 3, 6, 8]

result = all(num % 2 == 0 for num in numbers)
print(result)

# since 1 and 3 are odd, all() returns False
# all() returns True only if every number in the list satisfies the condition
````
