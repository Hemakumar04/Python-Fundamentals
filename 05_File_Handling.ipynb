# 05. File Handling in Python

This notebook covers text files, CSV, JSON, pickle, context managers, pathlib, os, shutil, validation, and directory operations with examples.

## 1. Reading Text Files

Reading a text file means opening it in read mode and processing its contents line by line or all at once.

### Real-World Example

A script loads a configuration file or a list of user emails from a text file at startup.

````python
with open('sample.txt', 'w', encoding='utf-8') as file:
    file.write('Line 1\nLine 2\nLine 3\n')

with open('sample.txt', 'r', encoding='utf-8') as file:
    content = file.read()

print('File content:')
print(content)
````

### Output

```
File content:
Line 1
Line 2
Line 3
```


## 2. Writing Text Files

Writing text files stores output, logs, or generated data on disk. Use write mode to create or overwrite files and append mode to add content.

### Real-World Example

A data processing pipeline writes processed text or reports to a file for later review.

````python
with open('output.txt', 'w', encoding='utf-8') as file:
    file.write('Report generated on 2026-08-05\n')
    file.write('Status: Success\n')

with open('output.txt', 'a', encoding='utf-8') as file:
    file.write('Additional update appended.\n')

with open('output.txt', 'r', encoding='utf-8') as file:
    print(file.read())
````

### Output

```
Report generated on 2026-08-05
Status: Success
Additional update appended.
```


## 3. CSV Files

CSV files are a common format for tabular data. Python's `csv` module makes it easy to read and write rows.

### Real-World Example

Exporting spreadsheet data as CSV for import into analytics tools or databases.

````python
import csv

rows = [['name', 'email'], ['Alice', 'alice@example.com'], ['Bob', 'bob@example.com']]
with open('contacts.csv', 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerows(rows)

with open('contacts.csv', 'r', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        print(row)
````

### Output

```
['name', 'email']
['Alice', 'alice@example.com']
['Bob', 'bob@example.com']
```


## 4. JSON Files

JSON is a text format for structured data. Use the `json` module to serialize Python objects to JSON and deserialize them back.

### Real-World Example

A web service writes configuration or API responses to JSON files for interoperability.

````python
import json

data = {'project': 'File Handling', 'version': 1.0, 'items': [1, 2, 3]}
with open('data.json', 'w', encoding='utf-8') as jsonfile:
    json.dump(data, jsonfile, indent=2)

with open('data.json', 'r', encoding='utf-8') as jsonfile:
    loaded = json.load(jsonfile)

print(loaded)
````

### Output

```
{'project': 'File Handling', 'version': 1.0, 'items': [1, 2, 3]}
```


## 5. Pickle Files

Pickle stores Python objects in binary form. It is useful for quick persistence of Python-specific data structures.

### Real-World Example

A machine learning workflow saves a trained model or preprocessed dataset to a pickle file.

````python
import pickle

payload = {'numbers': [10, 20, 30], 'message': 'pickle example'}
with open('payload.pkl', 'wb') as pklfile:
    pickle.dump(payload, pklfile)

with open('payload.pkl', 'rb') as pklfile:
    loaded_payload = pickle.load(pklfile)

print(loaded_payload)
````

### Output

```
{'numbers': [10, 20, 30], 'message': 'pickle example'}
```


## 6. Context Manager (`with`)

The `with` statement ensures resources like files are properly closed, even if errors occur. It is the recommended way to open files.

### Real-World Example

An ETL script uses `with` so it never leaves files open between processing steps.

````python
with open('context_example.txt', 'w', encoding='utf-8') as file:
    file.write('Context manager example\n')

with open('context_example.txt', 'r', encoding='utf-8') as file:
    print(file.readline().strip())

print('File closed:', file.closed)
````

### Output

```
Context manager example
File closed: True
```


## 7. `pathlib`

`pathlib` provides an object-oriented API for filesystem paths, making path manipulation more readable and cross-platform.

### Real-World Example

A project builds paths for data files in a way that works on Windows and Linux.

````python
from pathlib import Path

data_dir = Path('data_folder')
data_dir.mkdir(exist_ok=True)
file_path = data_dir / 'notes.txt'
file_path.write_text('Pathlib makes file paths easy.\n', encoding='utf-8')

print('File exists:', file_path.exists())
print('File name:', file_path.name)
````

### Output

```
File exists: True
File name: notes.txt
```


## 8. `os`

The `os` module offers lower-level file and directory operations, such as checking existence, listing contents, and removing files.

### Real-World Example

A deployment script inspects filesystem state and removes temporary files before starting a service.

````python
import os

current_dir = os.getcwd()
print('Current directory:', current_dir)
print('Contains sample.txt:', os.path.exists('sample.txt'))

if os.path.exists('temp_file.txt'):
    os.remove('temp_file.txt')
    print('Removed temp_file.txt')
else:
    print('No temp file to remove')
````

### Output

```
Current directory: <workspace path>
Contains sample.txt: True
No temp file to remove
```


## 9. `shutil`

The `shutil` module performs high-level file operations like copying, moving, and removing directory trees.

### Real-World Example

A backup process copies generated reports to an archive directory.

````python
import shutil
from pathlib import Path

src = Path('output.txt')
dst = Path('archive_output.txt')
if src.exists():
    shutil.copy(src, dst)
    print(f'Copied {src} to {dst}')
else:
    print('Source file not found for copy')
````

### Output

```
Copied output.txt to archive_output.txt
```
(or 'Source file not found for copy' if output.txt is missing)


## 10. File Validation

Validate files by checking existence, size, and extension before processing to avoid runtime errors.

### Real-World Example

A data ingestion script verifies that the input file exists and is not empty before loading its contents.

````python
from pathlib import Path

candidate = Path('contacts.csv')
if candidate.exists() and candidate.suffix == '.csv' and candidate.stat().st_size > 0:
    print('File is valid for processing:', candidate)
else:
    print('Invalid file:', candidate)
````

### Output

```
File is valid for processing: contacts.csv
```
(or 'Invalid file: contacts.csv' if the file is missing or empty)


## 11. Directory Operations

Directory operations include creating, listing, renaming, and deleting directories. These actions help organize files and manage workspace structure.

### Real-World Example

A build script creates output directories and cleans old artifacts before writing new build files.

````python
from pathlib import Path
import shutil

reports_dir = Path('reports')
reports_dir.mkdir(exist_ok=True)
print('Reports directory exists:', reports_dir.exists())

for child in reports_dir.iterdir():
    print('Child item:', child.name)

old_dir = Path('old_reports')
if old_dir.exists():
    shutil.rmtree(old_dir)
    print('Removed old_reports directory')
else:
    print('No old_reports directory to remove')
````

### Output

```
Reports directory exists: True
Child item: <file or folder names>
No old_reports directory to remove
```
