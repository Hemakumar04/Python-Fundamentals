## 1. try, except, else, and finally

Exceptions are errors that happen while a program is running. They can be handled so the program does not crash.

### try
**Explanation:** The try block contains code that might raise an exception.

**Example:**
```python
try:
    result = 10 / 2
    print(result)
```

**Output:**
```text
5.0
```

### except
**Explanation:** The except block handles the error if something goes wrong in the try block.

**Example:**
```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
```

**Output:**
```text
Cannot divide by zero
```

### else
**Explanation:** The else block runs only if no exception occurs in the try block.

**Example:**
```python
try:
    number = int("25")
except ValueError:
    print("Invalid number")
else:
    print("Conversion successful")
```

**Output:**
```text
Conversion successful
```

### finally
**Explanation:** The finally block always runs whether an exception occurs or not.

**Example:**
```python
try:
    print("Trying to open a file")
finally:
    print("This always executes")
```

**Output:**
```text
Trying to open a file
This always executes
```

````python
try:
    number = int("123")
except ValueError:
    print("That is not a valid integer")
else:
    print("Conversion was successful")
finally:
    print("This always runs")
````

## 2. raise

The raise statement is used to manually trigger an exception.

````python
age = -3

if age &lt; 0:
    raise ValueError("Age cannot be negative")
````

## 3. assert

The assert statement checks whether a condition is true. If it is false, an AssertionError is raised.

````python
password = "abc123"
assert len(password) &gt;= 8, "Password is too short"
````

## 4. Custom Exceptions

You can create your own exception classes when you want more specific error handling.

````python
class InvalidAgeError(Exception):
    pass

age = 15

try:
    if age &lt; 18:
        raise InvalidAgeError("You must be at least 18")
    else:
        print("Access granted")
except InvalidAgeError as e:
    print("Handled custom exception:", e)
````

## 5. Built-in Exceptions

Below are common built-in exceptions, when they occur, and how to handle them.

### 5.1 ValueError

Occurs when a function receives a value of the correct type but an invalid content.

````python
try:
    number = int("hello")
except ValueError as e:
    print("Handled ValueError:", e)
````

### 5.2 TypeError

Occurs when an operation is applied to an object of an inappropriate type.

````python
try:
    result = 10 + "5"
except TypeError as e:
    print("Handled TypeError:", e)
````

### 5.3 IndexError

Occurs when you try to access a list or tuple element with an index that is out of range.

````python
try:
    numbers = [1, 2, 3]
    print(numbers[5])
except IndexError as e:
    print("Handled IndexError:", e)
````

### 5.4 KeyError

Occurs when you try to access a dictionary key that does not exist.

````python
try:
    student = {"name": "Asha"}
    print(student["age"])
except KeyError as e:
    print("Handled KeyError:", e)
````

### 5.5 AttributeError

Occurs when an object does not have the attribute or method you are trying to use.

````python
try:
    text = "hello"
    print(text.uppercase())
except AttributeError as e:
    print("Handled AttributeError:", e)
````

### 5.6 NameError

Occurs when a variable or function name is not defined.

````python
try:
    print(unknown_variable)
except NameError as e:
    print("Handled NameError:", e)
````

### 5.7 FileNotFoundError

Occurs when a file you want to open does not exist.

````python
try:
    file = open("missing_file.txt", "r")
except FileNotFoundError as e:
    print("Handled FileNotFoundError:", e)
````

### 5.8 ZeroDivisionError

Occurs when you divide by zero.

````python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Handled ZeroDivisionError:", e)
````

### 5.9 ImportError

Occurs when an imported name cannot be found in the target module.

````python
try:
    from math import does_not_exist
except ImportError as e:
    print("Handled ImportError:", e)
````

### 5.10 ModuleNotFoundError

Occurs when Python cannot find the module you are trying to import.

````python
try:
    import this_module_does_not_exist
except ModuleNotFoundError as e:
    print("Handled ModuleNotFoundError:", e)
````
