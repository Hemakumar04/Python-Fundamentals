## Section 1 – Classes, Objects and Attributes

### Class
#### Definition

A class is a blueprint that defines the structure (attributes) and behaviour (methods) that its objects will have. The class itself does not hold real data - it only describes what any object built from it will contain.

#### Real-Life Example

A class is like an architectural blueprint for a house. The blueprint describes the rooms and doors, but nobody can live inside the blueprint itself.

#### Example

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def describe(self):
        return f"{self.brand} {self.model}"

print(type(Car))
```

### Object
#### Definition

An object is a real instance created from a class. While the class is the blueprint, the object is the actual thing built from it, with its own specific data.

#### Real-Life Example

If the class is a house blueprint, the object is the actual house built at a specific address, with real walls and a real door.

#### Example

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def describe(self):
        return f"{self.brand} {self.model}"

car1 = Car("Toyota", "Corolla")
car2 = Car("Honda", "Civic")

print(car1.describe())
print(car2.describe())
```

### Constructor (`__init__`)
#### Definition

The constructor is a special method, `__init__`, that Python automatically runs whenever a new object is created. It initialises the object's attributes with starting values.

#### Real-Life Example

The constructor is like the setup steps when unboxing a new phone - it walks through initial setup before the phone is ready to use.

#### Example

```python
class Student:
    def __init__(self, name, roll_number):
        print(f"Setting up Student object for {name}")
        self.name = name
        self.roll_number = roll_number

s1 = Student("Priya", 101)
print(s1.name, s1.roll_number)
```

### Instance Variables
#### Definition

Instance variables are attributes that belong to a specific object. Each object has its own independent copy, so changing one object's variable does not affect another object.

#### Real-Life Example

Instance variables are like each student's own notebook. Writing in one student's notebook does not change anyone else's.

#### Example

```python
class Wallet:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

w1 = Wallet("Asha", 500)
w2 = Wallet("Ravi", 1000)

w1.balance += 200
print(w1.owner, w1.balance)
print(w2.owner, w2.balance)
```

### Class Variables
#### Definition

Class variables are attributes shared by all instances of a class. They are defined inside the class body, not inside `__init__`, and every object sees the same value unless overridden.

#### Real-Life Example

A class variable is like a school's name printed on every student's ID card - every student shares that same value.

#### Example

```python
class Employee:
    company_name = "TechCorp"

    def __init__(self, name):
        self.name = name

e1 = Employee("Asha")
e2 = Employee("Ravi")

Employee.company_name = "TechCorp Global"
print(e1.company_name, e2.company_name)
```

## Section 2 – Methods

### Instance Methods
#### Definition

Instance methods take `self` as their first parameter, giving them access to that specific object's data. They define behaviour that operates on individual object data.

#### Real-Life Example

An instance method is like a personal action a specific student performs, e.g. 'submit my homework' - it acts on that student's own data only.

#### Example

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

acc = BankAccount("Meera", 1000)
print(acc.deposit(500))
```

### Static Methods
#### Definition

A static method, marked with `@staticmethod`, does not receive `self` or `cls` automatically. It behaves like a normal function grouped inside a class because it is logically related but does not need instance or class data.

#### Real-Life Example

A static method is like a public information desk in a company - it provides a general service that does not depend on any specific employee's personal data.

#### Example

```python
class MathUtils:
    @staticmethod
    def add(a, b):
        return a + b

print(MathUtils.add(3, 4))
```

### Class Methods
#### Definition

A class method, marked with `@classmethod`, receives `cls` (the class itself) as its first parameter. It is often used to create alternative constructors that build objects in different ways.

#### Real-Life Example

A class method is like a factory's head office issuing a company-wide policy - it operates on the company as a whole, not on one employee.

#### Example

```python
class Pizza:
    def __init__(self, toppings):
        self.toppings = toppings

    @classmethod
    def margherita(cls):
        return cls(["tomato", "mozzarella", "basil"])

p1 = Pizza.margherita()
print(p1.toppings)
```

### Properties (`@property`)
#### Definition

The `@property` decorator lets a method be accessed like a plain attribute, while still running code behind the scenes, commonly used for computed values or validation.

#### Real-Life Example

A property is like a thermostat display showing the current temperature - it looks like a simple number, but it's actually being measured behind the scenes.

#### Example

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height

r = Rectangle(4, 5)
print(r.area)
```

## Section 3 – Encapsulation and Abstraction

### Encapsulation
#### Definition

Encapsulation means bundling data and the methods that operate on it inside a class, while restricting direct outside access. A leading underscore signals internal use, and a double leading underscore triggers name mangling.

#### Real-Life Example

Encapsulation is like a medicine capsule - the ingredients are sealed inside, and we only interact with the capsule from the outside, not by touching the raw powder directly.

#### Example

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

acc = BankAccount("Asha", 1000)
acc.deposit(500)
print(acc.get_balance())
```

### Abstraction
#### Definition

Abstraction means exposing only the essential details to the user of a class while hiding complex implementation details, often achieved with the `abc` module and abstract methods that subclasses must implement.

#### Real-Life Example

Driving a car is abstraction in action - we just use the steering wheel and pedals without needing to understand the engine's internal mechanics.

#### Example

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class UpiPayment(PaymentMethod):
    def pay(self, amount):
        return f"Paid {amount} using UPI"

print(UpiPayment().pay(500))
```

## Section 4 – Inheritance

### Inheritance
#### Definition

Inheritance allows a new class to reuse the attributes and methods of an existing class, while also adding or overriding functionality. It models 'is-a' relationships.

#### Real-Life Example

A child inherits traits from a parent, like eye colour, but can also develop their own unique traits. A subclass works the same way with a parent class.

#### Example

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        return "Some generic animal sound"

class Dog(Animal):
    def make_sound(self):
        return "Woof!"

d = Dog("Rex")
print(d.name)
print(d.make_sound())
```

### Multiple Inheritance
#### Definition

Multiple inheritance is when a class inherits from more than one parent class at the same time, combining features from all of them.

#### Real-Life Example

A smartphone inherits 'camera' features from a camera device and 'calling' features from a telephone device, combining both into one object.

#### Example

```python
class Camera:
    def take_photo(self):
        return "Photo taken"

class Phone:
    def make_call(self):
        return "Calling..."

class SmartPhone(Camera, Phone):
    pass

s = SmartPhone()
print(s.take_photo())
print(s.make_call())
```

### Multilevel Inheritance
#### Definition

Multilevel inheritance is a chain of inheritance across more than two levels - a grandchild class inherits from a child class, which itself inherits from a parent class.

#### Real-Life Example

This is like a family tree: a grandchild inherits traits passed down through a parent, who inherited traits from the grandparent.

#### Example

```python
class Vehicle:
    def general_info(self):
        return "This is a vehicle"

class Car(Vehicle):
    def car_info(self):
        return "This is a car"

class SportsCar(Car):
    def sports_info(self):
        return "This is a sports car"

sc = SportsCar()
print(sc.general_info())
print(sc.car_info())
print(sc.sports_info())
```

### Hierarchical Inheritance
#### Definition

Hierarchical inheritance is when multiple child classes inherit from the same single parent class, sharing common functionality while each can extend it differently.

#### Real-Life Example

Several departments in a company all inherit the same company policies from head office, while each department also has its own specific rules.

#### Example

```python
class Employee:
    def __init__(self, name):
        self.name = name

class Developer(Employee):
    def work(self):
        return f"{self.name} is writing code"

class Designer(Employee):
    def work(self):
        return f"{self.name} is designing UI"

print(Developer("Ravi").work())
print(Designer("Asha").work())
```

## Section 5 – Polymorphism

### Polymorphism
#### Definition

Polymorphism means the same method name can behave differently depending on the object it is called on, letting code work generically across different object types that share a common interface.

#### Real-Life Example

Pressing 'play' on a device is polymorphic - pressing play on a music player plays a song, while pressing play on a video player plays a video.

#### Example

```python
class Cat:
    def speak(self):
        return "Meow"

class Duck:
    def speak(self):
        return "Quack"

for animal in [Cat(), Duck()]:
    print(animal.speak())
```

### Method Overriding
#### Definition

Method overriding happens when a subclass provides its own implementation of a method that already exists in its parent class, replacing the parent's version for that subclass.

#### Real-Life Example

A branch office may override a general head-office policy with a more specific local policy, while still technically being 'the same policy'.

#### Example

```python
class Shape:
    def area(self):
        return "Area not defined"

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

print(Circle(3).area())
```

### Method Overloading (Python Approach)
#### Definition

Python does not support true method overloading like Java or C++. Instead, overloading-like behaviour is achieved using default arguments or `*args`/`**kwargs` inside a single method.

#### Real-Life Example

Instead of several different-sized cups for different amounts, Python uses one adjustable cup that can expand to hold whatever is poured in.

#### Example

```python
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c

calc = Calculator()
print(calc.add(5, 10))
print(calc.add(5, 10, 15))
```

## Section 6 – Class Relationships

### Composition
#### Definition

Composition is a 'has-a' relationship where one class contains an instance of another class, and the contained object's lifetime is tightly tied to the container.

#### Real-Life Example

A car has an engine. The engine is built specifically as part of that car and doesn't meaningfully exist independently of it.

#### Example

```python
class Engine:
    def __init__(self, horsepower):
        self.horsepower = horsepower

    def start(self):
        return f"Engine with {self.horsepower}HP starting"

class Car:
    def __init__(self, brand, horsepower):
        self.brand = brand
        self.engine = Engine(horsepower)

car = Car("Toyota", 150)
print(car.engine.start())
```

### Aggregation
#### Definition

Aggregation is also a 'has-a' relationship, but a looser one - the contained object can exist independently of the container and could be shared across multiple containers.

#### Real-Life Example

A university has professors, but professors exist independently of any one university and could work elsewhere.

#### Example

```python
class Professor:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, name, professor):
        self.name = name
        self.professor = professor

prof = Professor("Dr. Iyer")
cs_dept = Department("Computer Science", prof)
print(cs_dept.professor.name)
```

### Association
#### Definition

Association is a general 'uses-a' relationship between two independent classes, where objects interact without either owning or being a structural part of the other.

#### Real-Life Example

A teacher and a student are associated - a teacher teaches a student - but neither one owns the other.

#### Example

```python
class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self, student):
        return f"{self.name} is teaching {student.name}"

class Student:
    def __init__(self, name):
        self.name = name

print(Teacher("Mr. Sharma").teach(Student("Priya")))
```

## Section 7 – Magic Methods

### Magic (Dunder) Methods
#### Definition

Magic methods are special, predefined methods surrounded by double underscores that Python calls automatically for certain operations like printing, addition, or equality checks.

#### Real-Life Example

Magic methods are like universal remote buttons - defining `__add__` lets an object respond to the `+` button the same way built-in numbers do.

#### Example

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)
```

### `super()`
#### Definition

`super()` gives access to methods defined in a parent class from within a subclass, most commonly used to call the parent's constructor before adding subclass-specific setup.

#### Real-Life Example

`super()` is like a new employee following the standard onboarding checklist first, before completing extra department-specific steps.

#### Example

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed

d = Dog("Rex", "Labrador")
print(d.name, d.breed)
```

### `__str__()`
#### Definition

`__str__` defines the readable string representation of an object, used automatically by `print()` and `str()`.

#### Real-Life Example

`__str__` is like the friendly, plain-English label on a product package aimed at customers.

#### Example

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f'"{self.title}" by {self.author}'

print(Book("1984", "George Orwell"))
```

### `__repr__()`
#### Definition

`__repr__` defines the official, developer-facing string representation of an object, used by `repr()` and as a fallback for `print()` if `__str__` is not defined.

#### Real-Life Example

`__repr__` is like a product's technical spec sheet used by engineers, as opposed to `__str__`'s friendly customer-facing label.

#### Example

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book(title={self.title!r}, author={self.author!r})"

print(repr(Book("1984", "George Orwell")))
```

### `__len__()`
#### Definition

`__len__` lets a custom object respond to Python's built-in `len()` function by defining what length means for that object.

#### Real-Life Example

`__len__` is like a checkout scanner reading how many items are in a basket - `len()` just asks the object how many things it contains.

#### Example

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

p = Playlist(["Song A", "Song B", "Song C"])
print(len(p))
```

## Section 8 – Bonus Topics

### Dataclasses (Bonus)
#### Definition

The `@dataclass` decorator automatically generates common boilerplate methods, like `__init__`, `__repr__`, and `__eq__`, for classes that mainly hold data.

#### Real-Life Example

A dataclass is like a pre-printed form with fields already laid out - we just fill in the values instead of designing the entire form from scratch.

#### Example

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p1 = Point(2, 3)
p2 = Point(2, 3)

print(p1)
print(p1 == p2)
```

### Singleton Pattern (Bonus)
#### Definition

The Singleton pattern ensures a class can have only one instance throughout the program, providing one global point of access to it.

#### Real-Life Example

A Singleton is like a country's president's office - no matter how many times someone tries to create a new one, there is only ever one real office.

#### Example

```python
class ConfigManager:
    _instance = None

    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.settings = {}
        return cls._instance

config1 = ConfigManager()
config2 = ConfigManager()
config1.settings["theme"] = "dark"

print(config2.settings)
print(config1 is config2)
```
