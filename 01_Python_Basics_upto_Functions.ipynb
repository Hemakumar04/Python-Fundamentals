## Section 1 – Basic Concepts

### Variables
A variable is a named place to store a value. It is like a box with a label that keeps information. The value inside the variable can change whenever we want.

Example:

age = 24

* `age` – variable
* `24` – value stored in the variable

We can use `age` instead of writing the number everywhere.

```python
# Creating and using variables
age = 24
name = "Ravi"

print(name, "is", age, "years old")

# Reassigning a variable
age = 25
print("After birthday, age =", age)
```

### Data Types
A data type tells Python what kind of information is stored in a variable. Different kinds of information need different data types. It tells whether the value is a number, text, decimal number, or True/False.

Some common data types are:

* Integer (`int`) – Whole numbers
* Float (`float`) – Decimal numbers
* String (`str`) – Text
* Boolean (`bool`) – True or False

```python
# Checking data types
roll_no = 21
height = 5.6
city = "Chennai"
is_present = True

print(type(roll_no))
print(type(height))
print(type(city))
print(type(is_present))
```

### Type Casting
Type casting means changing one data type into another. It changes data from one form to another, like text to a number.

Example:

* `"25"` (str)
* `25` (int)

Both represent twenty-five, but Python treats them differently.

```python
# Converting string to int and float
value = "25"

num_int = int(value)
num_float = float(value)

print(num_int, type(num_int))
print(num_float, type(num_float))

# Converting int to string
score = 90
score_str = str(score)
print(score_str, type(score_str))
```

### Operators
Operators are symbols that perform actions on values. They are used for calculations, comparisons, or checking conditions.

Some examples are:

Arithmetic operators:
(+) Add
(-) Subtract
(*) Multiply
(/) Divide

Comparison Operators:
(>) Greater than
(<) Less than
(==) Equal to

```python
# Arithmetic operators
a = 15
b = 4

print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)

# Comparison operators
print(a > b)
print(a < b)
print(a == b)
```

## Section 2 – Control Flow

### if / elif / else
`if`, `elif`, and `else` help Python make decisions.

* `if` checks the first condition.
* `elif` checks another condition if the first one is false.
* `else` runs when none of the conditions are true.

Only one matching block is executed.

Example

Imagine traffic lights.

* If the light is Green → Go
* Else if the light is Yellow → Slow
* Else → Stop

Python checks each condition one by one. As soon as one condition is true, it performs that action.

```python
# Grading a student's marks
marks = 78

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

### for loop
#### Definition

A for loop is a control statement used to repeat a block of code for each item in a sequence or for a specific number of times.

### Real-Life Example

Imagine a teacher checking attendance for 30 students.

Instead of saying each student's name manually in separate actions, the teacher goes through the class list one student at a time until everyone is checked.

A for loop works the same way—it automatically repeats the same task for every item in a sequence.
### Example

```python
# Print squares of numbers from 1 to 5
for i in range(1, 6):
    print(i, "*", i, "=", i * i)
```

### while loop

#### Definition

A while loop is a control statement that repeats a block of code as long as a specified condition is True.

### Real-Life Example

Imagine filling a water tank.

While the tank is not full, keep filling it.

When the tank becomes full, stop.

The loop continues until the condition becomes False.

### Example

```python
# Countdown using while loop
count = 5

while count > 0:
    print(count)
    count -= 1

print("Liftoff!")
```

## Break
### Definition

The `break` statement is used to immediately terminate a loop, even if the loop has not completed all its iterations.

### Real-Life Example

Imagine you're searching for your friend in a crowd.

As soon as you find them, you stop searching.

You don't continue looking through the remaining people.

That's exactly what `break` does.

### Example

```python
# Stop the loop once the target roll number is found
roll_numbers = [101, 102, 103, 104, 105]
target = 103

for roll in roll_numbers:
    if roll == target:
        print("Student found:", roll)
        break
else:
    print("Not in the list")
```

# Continue
## Definition

The `continue` statement is used to skip the current iteration of a loop and continue with the next iteration.

## Real-Life Example

Imagine a teacher checking homework.

One student forgot the homework.

The teacher skips that student and continues checking the rest of the class.

The checking doesn't stop—it simply skips one student.

### Example

```python
# Skip absent students (marked as None) while printing attendance
students = ["Aishu", None, "Divya", None, "Kavi"]

for student in students:
    if student is None:
        continue
    print(student, "is present")
```

## Section 3 – Collections

# Collections
### Definition

A collection is a data type that can store multiple values in a single variable. Instead of creating many variables, we can group related data together using collections.

### Real-Life Example

Think of a shopping basket.

Instead of carrying each item separately, you put all the items into one basket.

Similarly, a collection stores multiple values in one place.

# List
### Definition

A list is an ordered, mutable collection that can store multiple items. It allows duplicate values and items can be added, removed, or modified.

Ordered → Items keep their position.
Mutable → Items can be changed after creation.

### Real-Life Example

Think of your shopping list.

Milk
Bread
Eggs
Rice

### You can:

Add a new item.
Remove an item.
Change an item.

Just like a Python list.

### Example:

fruits = ["Apple", "Banana", "Mango"]

```python
# Create a list of students
students = ["Aishu", "Brindha", "Celine", "Divya"]
print(students)

# Add a new student
students.append("Arthi")
print(students)

# Remove a student
students.remove("Celine")
print(students)

# Change a student's name using index
students[1] = "Barathi"
print(students)

# Access a particular student
print(students[0])

# Total number of students
print(len(students))

# Sort in alphabetical order
students.sort()
print(students)
```

## Tuple
### Definition

A tuple is an ordered, immutable collection that stores multiple items. Once created, its values cannot be changed.

Ordered → Items keep their position.
Immutable → Items cannot be modified.

### Real-Life Example

Think of your Date of Birth.

### Example:

15-08-2000

Your date of birth doesn't change.

Similarly, once a tuple is created, its values remain the same.

### Easy way to remember:

List ([]) → Can add, remove, and change items.

Tuple (()) → Can only store, access, count, and search items. It cannot be modified after creation.

### Example:

birth_date = (15, 8, 2000)

```python
# Create tuple and access elements
fruits = ("apple", "banana", "mango", "orange", "papaya")

print(fruits)
print(fruits[0])          # indexing
print(fruits[-1])         # negative indexing
print(fruits[1:4])        # slicing
print(len(fruits))        # length
print("apple" in fruits)  # membership

# concatenation
colours = ("red", "orange", "yellow")
print(fruits + colours)

# tuple unpacking
student = ("Aishu", 24, "Chennai")
name, age, city = student
print(name, age, city)

# count elements
print(fruits.count("orange"))
```

```python
# Immutable - this raises an error
fruits = ("apple", "banana", "mango")
fruits[1] = "orange"
```

## Set
### Definition

A set is an unordered collection of unique items. It automatically removes duplicate values.

Unordered → Items do not have a fixed position.
Unique → Duplicate values are not allowed.

### Real-Life Example

Imagine a classroom.

Five students write their favorite fruit.

Apple
Mango
Apple
Orange
Mango

If the teacher wants only the unique fruits, the result becomes:

Apple
Mango
Orange

This is exactly how a set works.

### Example:

```python
# only unique items
numbers = {1, 2, 2, 3, 4, 4}
print(numbers)

# adding items
fruits = {"apple", "banana"}
fruits.add("mango")
print(fruits)

# removing items
fruits.remove("banana")
print(fruits)

# union of 2 sets
frt = {"apple", "orange", "banana"}
clr = {"red", "orange", "green"}
print(frt.union(clr))

# intersection of 2 sets
print(frt.intersection(clr))
```

## Dictionary
### Definition

A dictionary is a collection that stores data as key-value pairs. Each key is unique and is used to access its corresponding value.

### Real-Life Example

Think of an English dictionary.

Word      Meaning
Apple     A fruit
Book      A collection of pages

The word is the key.

The meaning is the value.

Similarly,

student = {
    "Name": "Nanda",
    "Age": 25
}

Here,

Name → Key
Nanda → Value

```python
student = {
    "name": "Aishu",
    "age": 24
}

print(student)

# access values
print(student["name"])
print(student["age"])

# add a key-value pair
student["city"] = "Chennai"
print(student)

# remove an item
student.pop("city")
print(student)
```

## List Comprehension
### Definition

List comprehension is a short and simple way to create a new list by writing the loop and optional condition in a single line.

### Real-Life Example

Imagine a teacher writing only the names of students who passed an exam.

Instead of writing a long process, the teacher directly creates a new list containing only the passed students.

Similarly, list comprehension creates a new list in one line.

### Example:

```python
marks = [45, 88, 32, 91, 67]

# list of marks that are passing (>=50)
passed = [m for m in marks if m >= 50]
print(passed)

# lengths of words
words = ["cat", "elephant", "dog"]
lengths = [len(word) for word in words]
print(lengths)
```

## Dictionary Comprehension
### Definition

Dictionary comprehension is a short and simple way to create a new dictionary by writing the loop and optional condition in a single line.

### Real-Life Example

Imagine a teacher preparing a report card.

Instead of writing each student's name and mark one by one, the teacher quickly creates the entire report.

Similarly, dictionary comprehension creates key-value pairs efficiently.

### Example:

```python
fruits = ["apple", "banana", "mango"]

fruit_length = {fruit: len(fruit) for fruit in fruits}
print(fruit_length)
```

## Section 4 – Functions

## Function
### Definition

A function is a reusable block of code designed to perform a specific task. Instead of writing the same code multiple times, you can write it once inside a function and call it whenever needed.

### Real-Life Example

Think of a coffee machine.

Whenever you press the "Coffee" button, the machine prepares coffee using the same steps every time.

Similarly, a function performs the same task whenever it is called.

### Example

```python
# function to check odd or even
def even_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(even_odd(7))
print(even_odd(2))
```

## Arguments
### Definition

Arguments are the values passed to a function when it is called. They provide the information the function needs to perform its task.

### Real-Life Example

Imagine ordering a pizza.

The pizza shop always asks:

What size?
Which toppings?

The information you give is like arguments passed to the function.

### Example

```python
def add(a, b):
    print(a + b)

add(10, 20)

# 10 is a since given first and 20 is b as it is given second like the arguments passed in add()
```

## Default Arguments
### Definition

Default arguments are parameters that already have a value assigned. If the caller does not provide a value for that parameter, the default value is used instead.

### Real-Life Example

Imagine ordering tea at a shop.

If you don't mention the sugar level, the shop gives you the "normal sugar" version by default.

If you specifically ask for "less sugar", it uses your choice instead.

### Example

```python
def greet(name, greeting="Hello"):
    print(greeting + ",", name + "!")

greet("Aishu")               # uses default greeting
greet("Ravi", "Good morning") # overrides the default
```

## Keyword Arguments
### Definition

Keyword arguments are arguments passed by parameter name, making it clear which value belongs to which parameter. The order of arguments doesn't matter.

Real-Life Example

Think of filling out an online application form.

Instead of writing values in order, you enter them under labeled fields like:

Name
Age
City

The labels identify the values.

### Example

```python
def add(a, b):
    print("Sum =", a + b)

add(b=20, a=10)

# here we have assigned the numbers using the respective parameters
```

## Variable-Length Arguments (`*args`, `**kwargs`)
### Definition

Variable-length arguments allow a function to accept any number of arguments.

`*args` collects multiple positional arguments into a tuple.
`**kwargs` collects multiple keyword arguments into a dictionary.

### Real-Life Example

Imagine a fruit basket.

Some days it contains:

Apple
Mango

Other days it contains:

Apple
Mango
Banana
Orange

The basket can hold any number of fruits.

Similarly, `*args` and `**kwargs` allow a function to receive any number of inputs.

#### Example using `*args`

```python
def largest(*numbers):
    print("Largest =", max(numbers))

largest(10, 50, 20, 40)
```

#### Example using `**kwargs`

```python
def details(**data):
    for key, value in data.items():
        print(key, ":", value)

details(name="Aishu", age=24, city="Chennai")
```

## Lambda Functions
### Definition

A lambda function is a small anonymous function written in a single line. It is commonly used for simple operations that do not require a full function definition.

### Real-Life Example

Imagine using a calculator to quickly add two numbers.

You don't need to write detailed steps—you simply press the buttons and get the result.

Similarly, a lambda function performs a simple task in one line.

### Example

```python
maximum = lambda a, b: a if a > b else b

print(maximum(15, 20))
```

## Recursive Functions
### Definition

A recursive function is a function that calls itself to solve a problem. It continues calling itself until a stopping condition (called the base case) is reached.

### Real-Life Example

Imagine standing between two mirrors.

You see your reflection repeating again and again.

It continues until the reflections become too small to notice.

Similarly, a recursive function keeps calling itself until it reaches the stopping condition.

### Example

```python
# factorial of a number
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```
