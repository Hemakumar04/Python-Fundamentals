### Tables

Tables organize rows and columns for entities such as customers, orders, and products.

**Real-World Example:** An e-commerce platform stores each order as a row in the orders table with columns for customer, amount, and status.

````python
CREATE TABLE customers (
    id INT PRIMARY KEY,
    name VARCHAR(100),
    email VARCHAR(255)
);
````

### Primary Key

A primary key uniquely identifies each row in a table and prevents duplicate entries.

**Real-World Example:** An orders table uses order_id as the primary key so each order can be referenced exactly.

````python
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    total_amount DECIMAL(10,2)
);
````

### Output

```
Query OK, 0 rows affected (0.02 sec)
```

### Foreign Key

A foreign key references another table and enforces relationships between related records.

**Real-World Example:** An order_items table references orders so each item belongs to an existing order.

````python
CREATE TABLE order_items (
    item_id INT PRIMARY KEY,
    order_id INT,
    product_id INT,
    quantity INT,
    FOREIGN KEY (order_id) REFERENCES orders(order_id)
);
````

### Output

```
Query OK, 0 rows affected (0.02 sec)
```

### Relationships

Relationships connect tables in one-to-one, one-to-many, or many-to-many patterns.

**Real-World Example:** A blog application stores authors and posts so each author can be linked to multiple posts.

````python
CREATE TABLE authors (
    id INT PRIMARY KEY,
    name VARCHAR(100)
);

CREATE TABLE posts (
    post_id INT PRIMARY KEY,
    author_id INT,
    title VARCHAR(200),
    FOREIGN KEY (author_id) REFERENCES authors(id)
);
````

### Output

```
Query OK, 0 rows affected (0.03 sec)
```

### Constraints

Constraints enforce rules such as NOT NULL, UNIQUE, and CHECK to keep data valid.

**Real-World Example:** A user table requires unique email addresses and valid status values.

````python
CREATE TABLE users (
    id INT PRIMARY KEY,
    email VARCHAR(255) NOT NULL UNIQUE,
    status VARCHAR(20) CHECK (status IN ('active', 'inactive'))
);
````

## Queries

Queries read and filter data from tables to answer business questions.

### SELECT

SELECT retrieves columns from one or more tables.

**Real-World Example:** A reporting dashboard retrieves customer names and emails for a mailing list.

````python
SELECT name, email FROM customers;
````

### Output

```
name | email
Alice | alice@example.com
Bob | bob@example.com
```

### WHERE

WHERE filters rows using conditions.

**Real-World Example:** A support team fetches only open tickets assigned to a particular agent.

````python
SELECT * FROM orders WHERE status = 'shipped';
````

### Output

```
order_id | status
101 | shipped
105 | shipped
```

### ORDER BY

ORDER BY sorts rows by one or more columns.

**Real-World Example:** A sales report sorts products by revenue descending to show the top sellers.

````python
SELECT product_name, revenue FROM product_sales ORDER BY revenue DESC;
````

### Output

```
product_name | revenue
Widget | 12000.00
Gadget | 8000.00
```

### GROUP BY

GROUP BY aggregates rows that share values in specific columns.

**Real-World Example:** A finance report groups sales by region to compute total revenue per region.

````python
SELECT region, SUM(amount) FROM sales GROUP BY region;
````

### Output

```
region | SUM(amount)
North | 5400.00
West | 7200.00
```

### HAVING

HAVING filters groups after aggregation.

**Real-World Example:** A summary query shows only regions with total revenue above a threshold.

````python
SELECT region, SUM(amount) AS total FROM sales GROUP BY region HAVING SUM(amount) > 5000;
````

### Output

```
region | total
West | 7200.00
```

**Real-World Example:** A marketing query returns unique country values from the customer list.

## Joins

Joins combine rows from related tables using matching keys.

### INNER JOIN

INNER JOIN returns rows that match in both tables.

**Real-World Example:** A report lists orders matched with customer names.

````python
SELECT o.order_id, c.name FROM orders o INNER JOIN customers c ON o.customer_id = c.id;
````

### Output

```
order_id | name
101 | Alice
102 | Bob
```

### LEFT JOIN

LEFT JOIN returns all rows from the left table and matching rows from the right table.

**Real-World Example:** A customer list shows all customers, including those with no orders.

````python
SELECT c.name, o.order_id FROM customers c LEFT JOIN orders o ON c.id = o.customer_id;
````

### Output

```
name | order_id
Alice | 101
Bob | 102
Carol | NULL
```

### RIGHT JOIN

RIGHT JOIN returns all rows from the right table and matching rows from the left table.

**Real-World Example:** A reconciliation query lists all orders and matched customer information, including orphan orders.

````python
SELECT o.order_id, c.name FROM orders o RIGHT JOIN customers c ON o.customer_id = c.id;
````

### Output

```
order_id | name
101 | Alice
102 | Bob
NULL | Carol
```

### FULL JOIN

FULL JOIN returns rows when there is a match in either table.

**Real-World Example:** A reconciliation view shows all customers and all orders, including unmatched rows on both sides.

````python
SELECT c.name, o.order_id FROM customers c FULL JOIN orders o ON c.id = o.customer_id;
````

### Output

```
name | order_id
Alice | 101
Bob | 102
Carol | NULL
NULL | 110
```

### CROSS JOIN

CROSS JOIN produces the Cartesian product of two tables.

**Real-World Example:** A forecasting query pairs every product with every region to estimate coverage scenarios.

````python
SELECT p.name, r.region FROM products p CROSS JOIN regions r;
````

### Output

```
name | region
Widget | North
Widget | West
Gadget | North
Gadget | West
```

## Advanced SQL

Advanced SQL features help build reusable, efficient, and expressive queries.

### UNION

UNION combines result sets from two queries and removes duplicates by default.

**Real-World Example:** A company merges two department email lists into a single contact list.

````python
SELECT email FROM sales_team UNION SELECT email FROM support_team;
````

### Output

```
email
alice@example.com
bob@example.com
carol@example.com
```

### CASE

CASE returns different values depending on conditions.

**Real-World Example:** A payroll report labels employees as Junior, Mid, or Senior based on salary.

````python
SELECT name, salary, CASE WHEN salary < 50000 THEN 'Junior' WHEN salary < 90000 THEN 'Mid' ELSE 'Senior' END AS level FROM employees;
````

### Output

```
name | salary | level
Alice | 48000 | Junior
Bob | 76000 | Mid
Carol | 120000 | Senior
```

### Subqueries

Subqueries nest one query inside another to filter or compute values.

**Real-World Example:** A report finds customers who placed more than one order using a subquery.

````python
SELECT name FROM customers WHERE id IN (SELECT customer_id FROM orders GROUP BY customer_id HAVING COUNT(*) > 1);
````

### Output

```
name
Bob
```

### Common Table Expressions (CTEs)

CTEs define temporary named result sets that simplify complex queries.

**Real-World Example:** A monthly revenue report uses a CTE to compute totals before filtering top months.

````python
WITH monthly_sales AS (SELECT DATE_TRUNC('month', sold_at) AS month, SUM(amount) AS total FROM sales GROUP BY DATE_TRUNC('month', sold_at)) SELECT month, total FROM monthly_sales WHERE total > 5000;
````

### Output

```
month | total
2026-07-01 | 7200.00
```

### Views

Views are named queries that act like virtual tables and simplify repeated logic.

**Real-World Example:** A billing team creates a view for active invoices and queries it directly.

````python
CREATE VIEW active_invoices AS SELECT invoice_id, customer_id, amount FROM invoices WHERE status = 'open';
SELECT * FROM active_invoices;
````

### Output

```
invoice_id | customer_id | amount
201 | 102 | 150.00
202 | 105 | 240.00
```

### Indexes

Indexes speed up searches and joins on frequently filtered columns.

**Real-World Example:** A large orders table adds an index on order_date for faster date-range queries.

````python
CREATE INDEX idx_orders_date ON orders(order_date);
````

### Window Functions

Window functions perform calculations across sets of related rows without collapsing the result set.

**Real-World Example:** A ranking query computes salesperson performance by region.

````python
SELECT name, region, amount, RANK() OVER (PARTITION BY region ORDER BY amount DESC) AS rank FROM sales;
````

### Output

```
name | region | amount | rank
Alice | North | 1200.00 | 1
Bob | North | 950.00 | 2
```
