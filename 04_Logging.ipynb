# 04. Logging in Python

This notebook explains Python logging with descriptions, real-world examples, code, and output for each topic.

## 1. What is Logging?

Logging is a way to record events that happen while software is running. It helps developers understand the flow of execution and diagnose issues by storing timestamped messages about program behavior.

### Real-World Example

A web application records requests, errors, and performance metrics so administrators can review what happened after a problem occurs.

```python
import logging

logging.basicConfig(level=logging.INFO)
logging.info('Application started')
logging.warning('Cache miss for user data')
logging.error('Failed to connect to database')
```

## 2. Why use Logging?

Logging provides a persistent history of program behavior, which is valuable for debugging, auditing, and monitoring. Unlike print statements, logs can be filtered by severity, routed to files, and structured consistently.

### Real-World Example

A service running in production uses logs to track failures, so an engineer can investigate crashes after the fact without rerunning the service.

```python
import logging

logger = logging.getLogger('app')
logger.setLevel(logging.DEBUG)
handler = logging.StreamHandler()
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)

logger.debug('Debugging information')
logger.info('User login succeeded')
logger.error('Unable to fetch configuration')
```

## 3. `print()` vs `logging`

`print()` just writes text to standard output, while `logging` provides severity levels, flexible output destinations, and structured information. Logging is suitable for production systems because it can be adjusted without changing code logic.

### Real-World Example

During development, a programmer may use `print()` for quick checks. In production, the same information should be logged so it can be reviewed later and correlated with other events.

```python
print('Loading configuration...')

import logging
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')
logging.info('Loading configuration...')
```

## 4. Log Levels

Log levels indicate the severity of a message. Common levels are DEBUG, INFO, WARNING, ERROR, and CRITICAL. Use lower levels for routine information and higher levels for serious issues.

### Real-World Example

An e-commerce site logs DEBUG for detailed request handling during development, INFO for completed orders, WARNING for retryable payment problems, and ERROR for failed transactions.

```python
import logging
logging.basicConfig(level=logging.DEBUG, format='%(levelname)s: %(message)s')
logging.debug('Checking user credentials')
logging.info('User request processed')
logging.warning('Disk space running low')
logging.error('Payment gateway timeout')
logging.critical('System out of memory')
```

## 5. Log Formatters

Formatters define how log records are presented. They can include timestamps, logger names, levels, file names, and custom context.

### Real-World Example

A microservices infrastructure uses a formatter that includes service name and request ID so logs from many services can be aggregated and searched.

```python
import logging
logger = logging.getLogger('service')
logger.setLevel(logging.INFO)
handler = logging.StreamHandler()
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)
logger.info('Service started successfully')
```

## 6. File Handlers

File handlers send logs to a file on disk. This preserves logs across program runs and allows later inspection or monitoring by external tools.

### Real-World Example

A backend service writes logs to a file so operations staff can review history and detect errors after an outage.

```python
import logging

logger = logging.getLogger('file_example')
logger.setLevel(logging.INFO)
file_handler = logging.FileHandler('example.log')
file_handler.setFormatter(logging.Formatter('%(levelname)s: %(message)s'))
logger.addHandler(file_handler)

logger.info('Writing this message to example.log')
logger.warning('This warning also goes to example.log')
print('Log file created: example.log')
```

## 7. Console Handlers

Console handlers send logs to standard output or standard error. They are useful for local development, containers, or command-line utilities.

### Real-World Example

A Docker container uses console logs so the orchestration platform can collect and display logs from stdout/stderr.

```python
import logging

logger = logging.getLogger('console_example')
logger.setLevel(logging.INFO)
console_handler = logging.StreamHandler()
console_handler.setFormatter(logging.Formatter('%(levelname)s - %(message)s'))
logger.addHandler(console_handler)
logger.info('This message appears in the console')
```

## 8. Rotating Logs (Bonus)

Rotating logs automatically archive old log files and start fresh ones when a size limit or time limit is reached. This prevents log files from growing indefinitely.

### Real-World Example

A long-running service rotates logs daily or when files exceed a threshold so disk usage stays under control while preserving recent history.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger('rotate_example')
logger.setLevel(logging.INFO)
rotating_handler = RotatingFileHandler('rotating.log', maxBytes=200, backupCount=3)
rotating_handler.setFormatter(logging.Formatter('%(levelname)s: %(message)s'))
logger.addHandler(rotating_handler)

for i in range(20):
    logger.info(f'Log entry {i}')

print('Rotating log file created: rotating.log with backups')
```

## 9. Logging Best Practices

- Use named loggers, not the root logger.
- Choose appropriate log levels.
- Avoid logging sensitive data.
- Configure handlers and formatters centrally.
- Use logging in libraries with `__name__`.
- Keep log messages clear and actionable.

### Real-World Example

A team configures a shared logging setup for all services so all logs have the same timestamp format, include service names, and are easy to search across systems.

```python
import logging

def setup_logging():
    formatter = logging.Formatter('%(asctime)s %(name)s %(levelname)s: %(message)s')
    handler = logging.StreamHandler()
    handler.setFormatter(formatter)

    root = logging.getLogger()
    root.setLevel(logging.INFO)
    root.addHandler(handler)

setup_logging()
logger = logging.getLogger('best_practice')
logger.info('Using a configured logger with formatter')
logger.warning('Avoid logging credentials directly')
```
